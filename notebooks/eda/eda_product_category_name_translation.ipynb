# EDA: `product_category_name_translation`

Разведочный анализ таблицы `product_category_name_translation` из базы данных Olist.


### Импорты

```python
import pandas as pd

import matplotlib.pyplot as plt

from sqlalchemy import create_engine
from sqlalchemy.engine import URL
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

### Конфигурация

```python
USERNAME = os.getenv("DB_USERNAME")
PASSWORD = os.getenv("DB_PASSWORD")
HOST = os.getenv("DB_HOST")
PORT = os.getenv("DB_PORT")
DATABASE = os.getenv("DB_NAME")

url = URL.create(
    drivername="postgresql+psycopg2",
    username=USERNAME,
    password=PASSWORD,
    host=HOST,
    port=PORT,
    database=DATABASE,
)
```

```python
connection = create_engine(url=url)
```

## Анализ данных


```python
read_sql_query = "SELECT * FROM product_category_name_translation"
df = pd.read_sql(sql=read_sql_query, con=connection)
```
